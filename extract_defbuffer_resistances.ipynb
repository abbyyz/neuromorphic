# Extract resistances from defbuffer CSV files

This notebook finds CSV files whose names start with `defbuffer` in the `new wired measurements` folder,
parses the instrument-style CSV (looks for the header row that contains `Reading` and `Value`),
computes resistances (R = V / I) while skipping zero/invalid currents, and writes the results to disk.

```python
from pathlib import Path
import csv
import json
import math
from typing import Tuple, List

# Folder containing the defbuffer CSVs (absolute path)
FOLDER = Path('/Users/abbyzhou/slac 26/new wired measurements')
PATTERN = 'defbuffer*.csv'

def read_iv_csv_like(csv_path: Path) -> Tuple[List[float], List[float]]:
    """Return (voltage_list, current_list) reading the CSV-like file.
    The function looks for a row that contains both the column names 'Reading' and 'Value'
    (case-insensitive) and then parses subsequent numeric rows.
    Current is expected under the 'Reading' column (A) and voltage under 'Value' (V).
    """
    with csv_path.open('r', encoding='utf-8', errors='replace', newline='') as handle:
        reader = csv.reader(handle)
        header_row_index = None
        header = None
        rows = []
        for i, row in enumerate(reader):
            lower = [c.lower() for c in row]
            if 'reading' in lower and 'value' in lower:
                header_row_index = i
                header = row
                break
            rows.append(row)

    if header is None:
        raise ValueError(f'Could not find header row with Reading and Value in {csv_path.name}')

    # Re-open and skip to header_row_index, then parse following lines
    voltage = []
    current = []
    with csv_path.open('r', encoding='utf-8', errors='replace', newline='') as handle:
        reader = csv.reader(handle)
        for _ in range(header_row_index + 1):
            next(reader, None)
        # Map header indices (case-insensitive match)
        header_cells = [c.strip() for c in header]
        try:
            current_index = [c.lower() for c in header_cells].index('reading')
            voltage_index = [c.lower() for c in header_cells].index('value')
        except ValueError:
            raise ValueError('Header found but could not map Reading/Value indices')

        for row in reader:
            if len(row) <= max(current_index, voltage_index):
                continue
            try:
                i = float(row[current_index])
                v = float(row[voltage_index])
            except Exception:
                continue
            current.append(i)
            voltage.append(v)

    return voltage, current
```

```python
# Iterate files, compute resistances (skip zero currents), and save results
files = sorted(FOLDER.glob(PATTERN))
results = []  # list of dicts: { 'file': name, 'resistances': [...] }
for f in files:
    try:
        voltage, current = read_iv_csv_like(f)
    except Exception as e:
        print(f'Warning: skipping {f.name}: {e}')
        continue
    resistances = []
    for v, i in zip(voltage, current):
        if i == 0 or math.isclose(i, 0.0, abs_tol=0.0) or math.isnan(i):
            continue
        r = v / i
        if math.isfinite(r):
            resistances.append(r)
    results.append({'file': f.name, 'resistances': resistances})

# Save results to JSON and CSV (resistances as semicolon-separated)
out_json = FOLDER / 'defbuffer_resistances.json'
out_csv = FOLDER / 'defbuffer_resistances.csv'
with out_json.open('w', encoding='utf-8') as handle:
    json.dump(results, handle, indent=2)
with out_csv.open('w', encoding='utf-8', newline='') as handle:
    writer = csv.writer(handle)
    writer.writerow(['file', 'resistances_semicolon_separated'])
    for row in results:
        writer.writerow([row['file'], ';'.join(f'{r:.6g}' for r in row['resistances'])])

print(f'Saved {out_json}
Saved {out_csv}')
```
